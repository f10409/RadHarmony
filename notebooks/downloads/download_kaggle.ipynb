# Download Kaggle datasets

Generic helper for fetching Kaggle competition / public datasets. Downloads land under
`KAGGLEHUB_CACHE` (set below) — the resolved path is what you pass as `base_image_dir`
(or nearby) when instantiating a dataset.

> **Note.** A plain Python script (`.py`) is a perfectly valid alternative to this notebook if you prefer — e.g. run `python notebooks/download_<dataset>.py` from CI or a shell. Use whichever fits your workflow; the logic is identical.

## Prerequisites

1. `pip install kagglehub` (auto-installed below if missing).
2. **Kaggle API token** (new `KGAT_` format) — create at <https://www.kaggle.com/settings> → API → "Create New API Token". The next cell will prompt for it if `KAGGLE_API_TOKEN` is not already exported in your shell.
3. If it's a competition, accept the rules on the competition page first.

````python
import os 
from pathlib import Path
print (Path.home()/ "path/to")
````

````python
import getpass
import os
from pathlib import Path

# Where downloads land. Defaults to ~/datasets; override by exporting KAGGLEHUB_CACHE before launching Jupyter.
#DATASETS_DIR = Path(os.environ.get("KAGGLEHUB_CACHE") or (Path.home() / "datasets"))
DATASETS_DIR =  Path.home()/ "path/to"
DATASETS_DIR.mkdir(parents=True, exist_ok=True)
os.environ["KAGGLEHUB_CACHE"] = str(DATASETS_DIR)

# Kaggle auth: new-style KGAT_ token via KAGGLE_API_TOKEN env var.
# Prompt if it's not already set — we don't want the token committed with the notebook.
if not os.environ.get("KAGGLE_API_TOKEN"):
    os.environ["KAGGLE_API_TOKEN"] = getpass.getpass("Paste Kaggle API token (KGAT_...): ")

print("Downloads will be cached under:", DATASETS_DIR)
print("KAGGLE_API_TOKEN set:", bool(os.environ.get("KAGGLE_API_TOKEN")))
````

## Download: RANZCR CLiP Catheter and Line Classification

~11 GB (JPEG chest X-rays). Accept the competition rules at
<https://www.kaggle.com/competitions/ranzcr-clip-catheter-line-classification/data>
before running this cell.

````python
import subprocess

path = kagglehub.competition_download(
    "ranzcr-clip-catheter-line-classification"
)
print("Path to competition files:", path)

# Make files group-readable/writable to match NAS convention
subprocess.run(["chmod", "-R", "g+rw", path], check=True)
print("Permissions set: g+rw")
````
